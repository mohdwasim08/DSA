## Largest Element

- **TC O(N)**

```python
arr = [3,2,1,5,2]

largest = arr[0]

for i in range(len(arr)):
    if arr[i] > largest:
        largest = arr[i]

print(largest)
```

## Second Largest Element

## **Better Solution**

- **TC O(2N)**


```python
arr = [1,2,4,7,7,5]
largest = arr[0]
slargest = -1
for i in arr:
    if i > largest:
        largest = i

for i in arr:
    if  i > slargest and i != largest:
        slargest = i

print('largest:',largest)        
print('second largest:',slargest)        
```

## **Optimal Solution**

- **TC O(N)**

```python
arr = [1,2,4,7,7,5]
largest = arr[0]
slargest = -1

for i in arr:
    if i > largest:
        slargest = largest
        largest = i
    elif i > slargest and i != largest:
        slargest = i

print(slargest)
```
