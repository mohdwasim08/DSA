*Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.*

```python
arr = [1,2,3,4,5,6,7]

temp = arr[len(arr)-1]
i = 0
for i in range(len(arr)-1,0,-1):
    arr[i] = arr[i-1]


arr[0] = temp
print(arr)
```
