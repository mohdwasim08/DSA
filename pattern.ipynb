```python
#19 Pattern
n = int(input('Enter the number:'))
# uoeer part
for i in range(n):
  # left stars
  for j in range(n-i):
    print('*', end='')
    #space
  for j in range(2*i):
      print(" ", end='')
    # right stars
  for j in range(n-i):
       print('*', end='')

  print()

# Lower part
for i in range(n):
  #left stars
  for j in range(i+1):
    print("*", end='')
  #spaces
  for j in range(2*(n-i-1)):
    print(" ", end='')
  #right stars
  for j in range(i+1):
    print("*", end='')

  print()






```

```python
#20th Pattern
n = int(input('Enter the number:'))
#upper part
for i in range(n):
  #left stars
  for j in range(i+1):
    print('*', end='')
  #spaces
  for j in range(2*(n-i-1)):
    print(" ", end='')
  #right stars
  for j in range(i+1):
    print('*', end='')
  print()

#lowe part
for i in range(n):
  #left stars
  for j in range(n-i):
    print('*', end='')
  #spaces
  for j in range(2*i):
    print(" ", end='')
  #right stars
  for j in range(n-i):
    print('*', end='')

  print()

```

```python
#21th Pattern
n = int(input('Enter the number:'))
for i in range(n):
  for j in range(n):
    if i == 0 or i == n-1 or j == 0 or j == n-1:
      print('*', end='')
    else:
      print(' ', end='')
  print()
```

```python
n = int(input())
revnum = 0
dup = n
while n > 0:
    lastdigit = n % 10
    revnum = revnum * 10 + lastdigit
    n = n // 10

if dup == revnum:
    print("True:", revnum)
else:
    print("False")
```
