## Array Hashing

```python
arr = [1,2,2,3,1,4,1,4]

# Precumputation 
hash = [0] * 13

for num in arr:
    hash[num] += 1

# Queries

q = [1,2,3,4,5,6,12]

for num in q:
    print(hash[num])
    
```

## Mapping 

```python
arr = [1,2,2,3,1,2]
map = {}

for num in arr:       
    if num in map:
        map[num] += 1
    else:
        map[num] = 1

print(map)
```

## Character Hashing

```python
s = 'abcdbceafe'

map = {}
for char in s:
    if char in map:
        map[char]+=1
    else:
        map[char]=1  

print(map)  
```

```python
arr = [1, 2, 2, 3, 1, 4, 1, 4]

# Precompute frequency
freq = [0] * 5

for num in arr:
    freq[num] += 1

# Find highest and lowest frequency
max_freq = 0
max_element = 0

min_freq = float('inf')
min_element = 0

for num in range(1, 5):

    # Highest frequency
    if freq[num] > max_freq:
        max_freq = freq[num]
        max_element = num

    # Lowest frequency
    if freq[num] < min_freq:
        min_freq = freq[num]
        min_element = num

print("Highest frequency element:", max_element)
print("Highest frequency:", max_freq)

print("Lowest frequency element:", min_element)
print("Lowest frequency:", min_freq)
```
