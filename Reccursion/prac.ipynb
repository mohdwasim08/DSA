```python
def f(i,n):
    if i > n:
        return
    print(i)
    f(i+1, n )
f(1,6)    
```

```python
def f(i,n):
    if i < 1:
        return
    print(i)
    f(i-1, n )
f(6,6) 
```

```python
def f(n):
    if n <= 1:
        return n
    last = f(n-1)
    s_last = f(n-2)
    return last + s_last
f(5)



```

```python
def f(i,sum):
    if i < 1:
        print(sum)
        return
    f(i-1, sum+i)
f(5,0)
```

```python
def f(n):
    if n == 0:
        return 0
    return n + f(n-1)
f(5)
```

```python
# Fabbonaci number
def f(n):
    if n<=1:
        return n
    return f(n-1) + f(n-2)
f(6)

# O(n)
```
