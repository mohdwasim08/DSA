## Reverse an Array

```python
# By double variable
n = int(input("size of array:"))
a = list(map(int,input().split()))

def f(l,r):
    
    # Base case
    if (l>=r):
        return
    a[l], a[r] = a[r], a[l]
    f(l+1, r-1)

f(0,n-1)
print(a)

```

```python
# By single variable 
n = int(input('size of array:'))
arr = list(map(int,input().split()))

def f(i):
    if (i >= n//2):
        return
    arr[i], arr[n-i-1] = arr[n-i-1], arr[i]
    f(i+1)

f(0)
print(arr)
```

## Checking the palindrome

```python
s = input('Enter a string:')
n = len(s)
def f(i):
    if (i>=n//2):
        return True
    if (s[i] != s[n-i-1]):
        return False
    return f(i+1)
f(0)
```

```python
s = input('Enter a string:')
n = len(s)
def f(l,r):
    if (l >= r):
        return True
    if (s[l] != s[r]):
        return False
    return f(l+1, r-1)
f(0,n-1)
```

```python
# Revision 

def f(n,arr,i):
    if i >= n//2:
        return
    arr[i],arr[n-i-1] = arr[n-i-1],arr[i]
    f(n, arr,i+1)
arr = [1, 2, 3, 4, 5]
f(5,arr,0)
print(arr)    
```
