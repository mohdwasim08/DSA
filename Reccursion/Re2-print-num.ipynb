## Q1. Print Name N times 

```python
n = int(input('enter the number :'))
i = 1
def f(i,n):

    #Base Case:
    if i>n :
        return
    print('Wasim')

    f(i+1, n)

f(i,n)     



#Recursion Treee


#(1,3)-->f(2,3)-->f(3,3)-->f(4,3)
```

**Recursion Treee**
f(1,3)-->f(2,3)-->f(3,3)-->f(4,3)

- T.C : O(n)
- S.C : O(n)

## Q3. Print linearly from N to 1

```python
n = int(input('enter the number:'))

def f(i,n):

    #Base case 
    if i<1:
        return
    print(i)
    f(i-1, n)

f(n,n)
  
```

**Recursion Tree**
- f(4,4)-->f(3,4)-->f(2,4)-->f(1,4)--f(0,4(return))

- T.C : O(n)
- T.C : O(n)

## Q4. Print linearly from 1 to N (But by Backtrack)

```python
n = int(input('enter the number:'))

def f(i,n):

    #Base case
    if i<1:
        return
    f(i-1,n)
    print(i)

f(n,n)
```

**Recursion Tree**
     

- f(3,3)-->f(2,3)-->f(1,3)-->f(0,3)



## Q5. Print from N to 1(By Backtrack)

```python
n = int(input('enter the number:'))
def f(i,n):

    #base case:
    if i>n:
        return
    f(i+1,n)
    print(i)
f(i,n)    
```
