## Q. print sum of n number 

```python
# Parameterised way

n = int(input('enter the number:'))
def f(i,sum):
    
    # Base case
    if i<1:
        print(sum)
        return
    f(i-1,sum+i)
f(n,0)        

    


```

```python
# Functional way
n = int(input('enter the number:'))
def f(n):
    if n==0:
        return 0
    return n + f(n-1)
f(n)
```

## Q Print the factorial of N numbers 

```python
n = int(input('enter the number:'))
def f(n):
    if n == 0:
        return 1
    return n * f(n-1)
f(n)
```

```python
n = int(input('enter the number:'))
def f(i,fact):
    if i<1:
        print(fact)
        return
    f(i-1, fact*i)

f(n,1)
```
